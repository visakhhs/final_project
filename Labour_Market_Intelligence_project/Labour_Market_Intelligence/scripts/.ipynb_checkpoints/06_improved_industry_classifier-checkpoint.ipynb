```python
# Load necessary csv files
import pandas as pd
import numpy as np
import re

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    f1_score
)


# PATHS
INPUT_FILE = Path(
    r"C:\Users\Visakh H S\Labour_Market_Intelligence\data\processed\master_jobs_industry.csv"
)

OUTPUT_FILE = Path(
    r"C:\Users\Visakh H S\Labour_Market_Intelligence\data\processed\master_jobs_industry_improved.csv"
)


# LOAD DATA
df = pd.read_csv(INPUT_FILE)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print("Rows:", len(df))
print("Columns:", df.columns.tolist())


# REQUIRED COLUMN CHECK
required_columns = [
    "job_id",
    "title",
    "description_clean",
    "industry"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

print("\nRequired columns found successfully.")

display(
    df[
        [
            "job_id",
            "title",
            "description_clean",
            "industry"
        ]
    ].head()
)
```

```python
# rule_classify()
def rule_classify(title):

    title = str(title).lower()

    matches = []

    for industry, patterns in INDUSTRY_RULES.items():

        for pattern in patterns:

            if re.search(pattern, title):

                matches.append(industry)
                break

    # Only return a rule when exactly one industry matches
    if len(matches) == 1:
        return matches[0]

    # Multiple conflicting rules → let ML decide
    return None


df["industry_rule"] = df["title"].apply(rule_classify)

print("Rule-based classifications:")
print(df["industry_rule"].notna().sum())

print("\nUnclassified by rules:")
print(df["industry_rule"].isna().sum())

print("\nRule distribution:")
display(
    df["industry_rule"]
    .value_counts()
)
```

```python
# inspect rule classifications
display(
    df[
        df["industry_rule"].notna()
    ][
        [
            "title",
            "industry",
            "industry_rule"
        ]
    ].sample(
        min(50, df["industry_rule"].notna().sum()),
        random_state=42
    )
)
```

```python
# create rule training data
rule_training = df[
    df["industry_rule"].notna()
].copy()

rule_training["training_industry"] = (
    rule_training["industry_rule"]
)

print(
    "High-confidence rule training jobs:",
    len(rule_training)
)

print("\nTraining distribution:")

display(
    rule_training["training_industry"]
    .value_counts()
)
```

```python
# existing labelled data
existing_training = df[
    (df["industry"] != "Other") &
    (df["industry"].notna()) &
    (df["industry"].astype(str).str.strip() != "")
].copy()

existing_training["training_industry"] = (
    existing_training["industry"]
)

print(
    "Existing labelled training jobs:",
    len(existing_training)
)
```

```python
# combine training data
training_df = pd.concat(
    [
        existing_training[
            [
                "classification_text",
                "training_industry"
            ]
        ],

        rule_training[
            [
                "classification_text",
                "training_industry"
            ]
        ]
    ],
    ignore_index=True
)

# Remove duplicate training examples
training_df = training_df.drop_duplicates(
    subset=["classification_text", "training_industry"]
)

print(
    "Final training examples:",
    len(training_df)
)

print("\nTraining labels:")
display(
    training_df["training_industry"]
    .value_counts()
)
```

```python
# TF-IDF
X_text = training_df["classification_text"]
y = training_df["training_industry"]

X_train, X_test, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True,
    max_features=100000
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training matrix:", X_train_tfidf.shape)
print("Testing matrix:", X_test_tfidf.shape)
```

```python
# Logistic Regression
classifier = LogisticRegression(
    max_iter=2000,
    class_weight="balanced"
)

classifier.fit(
    X_train_tfidf,
    y_train
)

print("Industry classifier trained successfully.")
```

```python
# validation
y_pred = classifier.predict(X_test_tfidf)

accuracy = accuracy_score(
    y_test,
    y_pred
)

macro_f1 = f1_score(
    y_test,
    y_pred,
    average="macro"
)

weighted_f1 = f1_score(
    y_test,
    y_pred,
    average="weighted"
)

print("INDUSTRY CLASSIFIER VALIDATION")

print("Accuracy:", round(accuracy, 4))
print("Macro F1:", round(macro_f1, 4))
print("Weighted F1:", round(weighted_f1, 4))

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
# classify all 19,207 jobs
all_tfidf = vectorizer.transform(
    df["classification_text"]
)

ml_predictions = classifier.predict(
    all_tfidf
)

ml_probabilities = classifier.predict_proba(
    all_tfidf
)

ml_confidence = ml_probabilities.max(axis=1)

df["industry_ml"] = ml_predictions
df["industry_ml_confidence"] = ml_confidence

print("ML classification completed.")
```

```python
# hybrid classification
def hybrid_industry(row):

    rule = row["industry_rule"]
    ml = row["industry_ml"]
    confidence = row["industry_ml_confidence"]
    existing = row["industry"]

    
# 1. High-confidence title rule
    if pd.notna(rule):
        return rule


# 2. High-confidence ML prediction
    if confidence >= 0.65:
        return ml


# 3. Keep existing reliable classification
    if (
        pd.notna(existing)
        and str(existing).strip() != ""
        and existing != "Other"
    ):
        return existing

# 4. Otherwise retain Other
    return "Other"


df["industry_improved"] = df.apply(
    hybrid_industry,
    axis=1
)

print("Hybrid classification completed.")
```

```python
# SECOND-PASS INDUSTRY CLASSIFICATION FOR "OTHER"
print("=" * 70)
print("SECOND-PASS CLASSIFICATION OF OTHER JOBS")
print("=" * 70)


# Industry-specific TITLE keywords
SECOND_PASS_RULES = {

    "Manufacturing": [
        "manufacturing",
        "manufacturing technician",
        "production operative",
        "production technician",
        "production manager",
        "production supervisor",
        "machine operator",
        "machine technician",
        "moulding technician",
        "molding technician",
        "quality inspector",
        "quality technician",
        "assembly operative",
        "assembly technician",
        "process technician",
        "factory operative",
        "factory worker",
        "fabricator",
        "welder"
    ],

    "Transport and Logistics": [
        "class 1 driver",
        "class 2 driver",
        "hgv driver",
        "lgv driver",
        "lorry driver",
        "truck driver",
        "delivery driver",
        "van driver",
        "warehouse operative",
        "warehouse worker",
        "warehouse manager",
        "forklift driver",
        "forklift operator",
        "reach truck",
        "counterbalance",
        "loader",
        "side loader",
        "logistics coordinator",
        "logistics manager",
        "transport manager",
        "transport coordinator"
    ],

    "Finance and Banking": [
        "accounts payable",
        "accounts receivable",
        "accountant",
        "accounting assistant",
        "finance assistant",
        "finance manager",
        "financial analyst",
        "financial controller",
        "payroll administrator",
        "payroll officer",
        "bookkeeper",
        "credit controller",
        "credit analyst",
        "investment analyst",
        "private equity",
        "banking",
        "banker"
    ],

    "IT and Technology": [
        "software developer",
        "software engineer",
        "data analyst",
        "data scientist",
        "machine learning engineer",
        "ai engineer",
        "developer",
        "programmer",
        "devops",
        "cloud engineer",
        "systems administrator",
        "system administrator",
        "network engineer",
        "cybersecurity",
        "cyber security",
        "information security",
        "security analyst",
        "it support",
        "it technician",
        "database administrator"
    ],

    "Engineering": [
        "mechanical engineer",
        "electrical engineer",
        "civil engineer",
        "design engineer",
        "project engineer",
        "engineering manager",
        "engineering technician",
        "process engineer",
        "maintenance engineer",
        "structural engineer",
        "systems engineer",
        "chemical engineer",
        "nuclear engineer",
        "engineering consultant"
    ],

    "Construction": [
        "site manager",
        "site supervisor",
        "site engineer",
        "quantity surveyor",
        "construction manager",
        "construction worker",
        "bricklayer",
        "carpenter",
        "plumber",
        "electrician",
        "roofer",
        "groundworker",
        "building surveyor"
    ],

    "Healthcare": [
        "nurse",
        "nursing",
        "doctor",
        "pharmacist",
        "pharmacy",
        "healthcare assistant",
        "care assistant",
        "clinical",
        "healthcare",
        "medical",
        "therapist",
        "radiographer",
        "paramedic",
        "dentist",
        "dental"
    ],

    "Education": [
        "teacher",
        "teaching assistant",
        "lecturer",
        "professor",
        "tutor",
        "education",
        "school administrator",
        "academic",
        "head teacher",
        "teacher assistant"
    ],

    "Legal": [
        "solicitor",
        "barrister",
        "paralegal",
        "legal assistant",
        "legal secretary",
        "lawyer",
        "legal counsel",
        "compliance lawyer"
    ],

    "Human Resources": [
        "hr manager",
        "hr advisor",
        "hr adviser",
        "hr assistant",
        "human resources",
        "recruitment consultant",
        "recruiter",
        "talent acquisition",
        "people manager",
        "people advisor"
    ],

    "Marketing and Advertising": [
        "marketing manager",
        "marketing assistant",
        "marketing executive",
        "digital marketing",
        "social media manager",
        "content manager",
        "content creator",
        "seo specialist",
        "seo manager",
        "advertising manager",
        "brand manager"
    ],

    "Hospitality and Food Services": [
        "restaurant manager",
        "restaurant general manager",
        "chef",
        "head chef",
        "kitchen manager",
        "waiter",
        "waitress",
        "bartender",
        "bar manager",
        "hotel manager",
        "hotel receptionist",
        "hospitality",
        "catering"
    ],

    "Retail": [
        "sales assistant",
        "retail assistant",
        "retail manager",
        "store manager",
        "store assistant",
        "shop assistant",
        "shop manager",
        "retail supervisor",
        "cashier",
        "customer service assistant"
    ],

    "Automotive": [
        "automotive technician",
        "vehicle technician",
        "motor vehicle technician",
        "vehicle mechanic",
        "car mechanic",
        "automotive engineer",
        "motor mechanic",
        "vehicle mechanic"
    ],

    "Real Estate": [
        "estate agent",
        "property manager",
        "property administrator",
        "property consultant",
        "real estate",
        "letting agent",
        "lettings negotiator"
    ],

    "Insurance": [
        "insurance advisor",
        "insurance adviser",
        "insurance broker",
        "insurance underwriter",
        "insurance analyst",
        "claims handler",
        "claims advisor"
    ]
}



# Second-pass classifier
def second_pass_classify(title):
    
    title = str(title).lower().strip()

    if not title:
        return None

    matches = []

    for industry, keywords in SECOND_PASS_RULES.items():

        for keyword in keywords:

            keyword_pattern = r"\b" + re.escape(keyword.lower()) + r"\b"

            if re.search(keyword_pattern, title):
                matches.append(industry)
                break

    # Only classify if there is exactly one industry match
    if len(matches) == 1:
        return matches[0]

    # Ambiguous → leave unchanged
    return None


# Apply only to current "Other" jobs
other_mask = (
    df["industry"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("other")
)

df["second_pass_industry"] = None

df.loc[other_mask, "second_pass_industry"] = (
    df.loc[other_mask, "title"]
    .apply(second_pass_classify)
)


# Count results
second_pass_matches = (
    df["second_pass_industry"]
    .notna()
)

print(
    "\nSecond-pass classifications:",
    second_pass_matches.sum()
)

print(
    "Remaining Other:",
    other_mask.sum() - second_pass_matches.sum()
)

print("\nSecond-pass industry distribution:")

display(
    df.loc[
        second_pass_matches,
        "second_pass_industry"
    ].value_counts()
)
```

```python
# INSPECT SECOND-PASS CLASSIFICATIONS
second_pass_df = df[
    df["second_pass_industry"].notna()
].copy()

print(
    "Total second-pass classifications:",
    len(second_pass_df)
)

display(
    second_pass_df[
        [
            "job_id",
            "title",
            "industry",
            "second_pass_industry",
            "description_clean"
        ]
    ]
    .sample(
        n=min(100, len(second_pass_df)),
        random_state=42
    )
    .reset_index(drop=True)
)
```

```python
# APPLY SECOND-PASS CLASSIFICATIONS
before_other = (
    df["industry"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("other")
    .sum()
)

# Keep the original industry for auditability
df["industry_before_second_pass"] = df["industry"]

# Replace Other only where the second-pass classifier
# produced a confident single-industry match
second_pass_mask = df["second_pass_industry"].notna()

df.loc[
    second_pass_mask,
    "industry"
] = df.loc[
    second_pass_mask,
    "second_pass_industry"
]

after_other = (
    df["industry"]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("other")
    .sum()
)

print("=" * 70)
print("SECOND-PASS CLASSIFICATION APPLIED")
print("=" * 70)

print("\nClassifications applied:", second_pass_mask.sum())
print("Other before:", before_other)
print("Other after:", after_other)

print(
    "\nOther reduction:",
    before_other - after_other
)

print(
    "Other percentage after:",
    round(after_other / len(df) * 100, 2),
    "%"
)
```
